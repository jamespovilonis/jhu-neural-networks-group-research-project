# Dataset Metainfo

* Perhaps some plots or stats on dataset distribution and entries, or tables on data + data source

- This dataset has almost 90 million entries of data, spanning across 84 million daily gridded meteorlogical recordings over 7 years, 2.6 million, gridded fire detections/hotspots, and most valuable 1.4 million entries with fire detection information along with topography info and satelite weather information all in one entry.

- The size and length of this dataset allows for adequate size, history, fidelity, and accesibility.
    - Size --> almost 90 million entries, 1.4 million fused entries
    - history --> 2018 - 2025
    - Fidelity --> Less than 1x1 km spatial resolution
    - Accesbility --> fused entries allow for easy indexing and labeling for training and test



# Data Source: Brazil’s National Institute for Space Research (Instituto Nacional de Pesquisas Espaciais)

- Primary data label source  is INPE's BDQueimadas program
- INPE processes thermal infrared information from polar-orbiting and geostationary satellites
- When an active fire crosses a threshold temperature, INPE logs a 1x1 km hotspot coordinate, which will provide us truth labels for forest fires and asociated information.

# Data source: ERA5 and NASA STRM

- ERA5 is a large geospatial meteorlogical dataset combining physics based predictive models and historical data that provides hourly data points on a 0.25 x0.25 degree resolution
    - Includes entries like temperature, wind, precipitation, surface pressure, and dewpoint
    - This provides feature data that the model will use to build predictive mapping from meteorlogical factors to forest fire occurences

- NASA strm:
    - NASA STRM is a topgraphic dataset at about 30 meter resolution developed by flight to ground radar remote sensing.
    - Topographic data can either be used as a model input or as a preprocessing tool 

# Predictive Power

- This dataset lends an opportunity to model complex non linear relationships between wind speed, biome, topography, and moisture (dewpoint and precipitation).
- Basic intuitive and accepted relationships are a basis for our hypothesis that these factors can be used for fire prediction
    - Dryness allows for a greater likelihood of fire
    - Wind causes easier spread of fire
    - Biome can affect flammability
    

# Challenges associated with dataset

- Lack of negatives --> The datapoints in the fire detection categories are only positive samples of forest fires. In order to train a binary classication or even regression, we must have some set of negative samples in order to teach the model what wouldn't cause a fire. To do this, we can extrapolate sattelite meteorlogical information from times that a fire detection isn't listed to create a negative set.

- Complexity of data sources

- Language --> it is entire in portugese (?). While this seems trivial, we must translate labels and other metadata
